**4. Описова статистика**

Обчисліть (кожен пункт оцінюється в 0.5 бала):
- Середній розмір рахунку на всьому наборі даних
- Максимальні чайові
- Кількість пропущених значень у всіх колонках
- Скільки в наборі даних рахунків за обіди і за вечері

```python
import pandas as pd
import numpy as np

tips_data = pd.read_csv('tip.csv')

np_mean_bill = np.mean(tips_data['total_bill'] + tips_data['tip'])
pd_mean_bill = tips_data['total_bill'].mean() + tips_data['tip'].mean()
print(f'np Середнє значення total_bill + tip: {np_mean_bill}')
print(f'pd Середнє значення total_bill + tip: {pd_mean_bill}')
print('---------------------------------')
max_tip = tips_data["tip"].max()
print(f'Максимальні чайові: {max_tip}')
print('---------------------------------')
null_values_count = tips_data.isnull().sum()
print(f'Кількість пропущених значень у кожному стовпці:\n{null_values_count}')
print('---------------------------------')
dinner_count = tips_data[tips_data['time'] == 'Dinner'].shape[0]
lunch_count = tips_data[tips_data['time'] == 'Lunch'].shape[0]
print(f'Кількість записів для Dinner: {dinner_count}')
print(f'Кількість записів для Lunch: {lunch_count}')
```

**5. Статистика із групуванням**

Обчисліть (кожен пункт оцінюється в 0.5 бала):
- Середній рахунок за статтю (groupby `sex`)
- Середні чайові за днем тижня (groupby `day`)
- Скільки відвідувачів у кожен день

```python
import pandas as pd

tips_data = pd.read_csv('tip.csv')

def mean_by_group(data, group_col, value_col):
    return data.groupby(group_col)[value_col].mean()

def mean_by_group_agg(data, group_col, value_col):
    return data.groupby(group_col).agg({value_col: ['mean']})

mean_tip_by_gender = mean_by_group(tips_data, 'sex', 'tip')
agg_mean_tip_by_gender = mean_by_group_agg(tips_data, 'sex', 'tip')
print(f'Середні чайові за статтю:\n{mean_tip_by_gender}')
print(f'Середні чайові за статтю (agg):\n{agg_mean_tip_by_gender}')
print('---------------------------------')
mean_tip_by_day = mean_by_group(tips_data, 'day', 'tip')
agg_mean_tip_by_day = mean_by_group_agg(tips_data, 'day', 'tip')
print(f'Середні чайові за днем:\n{mean_tip_by_day}')
print(f'Середні чайові за днем (agg):\n{agg_mean_tip_by_day}')
print('---------------------------------')
total_visitors_by_day = tips_data.groupby('day')['size'].sum()
agg_total_visitors_by_day = tips_data.groupby('day').agg({'size': ['sum']})
print(f'Загальна кількість відвідувачів за днем:\n{total_visitors_by_day}')
print(f'Загальна кількість відвідувачів за днем (agg):\n{agg_total_visitors_by_day}')
```

**7. Сортування**

Відсортуйте датасет і виведіть результат сортування в кожному випадку окремо (кожен пункт оцінюється в 0.5 бала):
- За розміром рахунку (зростання)
- За чайовими (спадання)

Додатково виведіть топ-5 найбільших чайових використовуючи метод `nlargest`.

```python
import pandas as pd

tips_data = pd.read_csv('tip.csv')

tips_data.sort_values(by='total_bill', ascending=True, inplace=True)
print(f'Перші 5 рядків після сортування за total_bill (зростання):\n{tips_data.head()}')
tips_data.sort_values(by='tip', ascending=False, inplace=True)
print(f'Перші 5 рядків після сортування за tip (спадання):\n{tips_data.head()}')
print(f'Перші 5 рядків з найбільшими значеннями tip:\n{tips_data.nlargest(5, "tip")}')
```
