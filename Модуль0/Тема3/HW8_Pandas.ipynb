Tips - це датасет, який містить інформацію про рахунки та чайові в ресторані.

**3. Фільтрація даних**

Відфільтруйте повний набір даних та виведіть (кожен пункт оцінюється в 0.5 бала):
- Всі рахунки більше 20 доларів (`total_bill > 20`)
- Всі відвідувачі-жінки (`sex == 'Female'`)
- Рахунки на вечерю (`time == 'Dinner'`)
- Рахунки на вечерю з сумою чеку більше 30 доларів

Для виведення результатів фільтрації:
- зберегти результат у окрему змінну
- показати кількість записів у відфільтрованому наборі
- вивести перші 5 записів відфільтрованого набору.

```python
import pandas as pd

tips_data = pd.read_csv('tip.csv')

bill_mask = tips_data['total_bill'] > 20
bill_masked_data = tips_data[bill_mask]
print(f'Рядки з total_bill > 20:\n{bill_masked_data.head()}')
print(f'Кількість записів: {bill_masked_data.shape[0]}')
print('---------------------------------')
gender_mask = tips_data['sex'] == 'Female'
gender_masked_data = tips_data[gender_mask]
print(f'Рядки з sex == Female:\n{gender_masked_data.head()}')
print(f'Кількість записів: {gender_masked_data.shape[0]}')
print('---------------------------------')
time_mask = tips_data['time'] == 'Dinner'
time_masked_data = tips_data[time_mask]
print(f'Рядки з time == Dinner:\n{time_masked_data.head()}')
print(f'Кількість записів: {time_masked_data.shape[0]}')
print('---------------------------------')
total_bill_mask = tips_data['total_bill'] + tips_data['tip'] > 30
total_bill_masked_data = tips_data[total_bill_mask]
print(f'Рядки з total_bill + tip > 30:\n{total_bill_masked_data.head()}')
print(f'Кількість записів: {total_bill_masked_data.shape[0]}')
```
