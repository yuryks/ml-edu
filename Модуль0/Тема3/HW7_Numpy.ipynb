**1. Створення масивів**

Створіть (кожен пункт оцінюється в 0.5 бала):
- Масив зі списку `[5, 15, 25, 35, 45]`
- Масив з нулів розміром 7
- Масив з одиниць розміром 6
- Масив з числами від 10 до 19 використовуючи `np.arange()`

```python
import numpy as np

list = [1, 2, 3, 4, 5]
array = np.array(list)
zeros = np.zeros((7,))
ones = np.ones((2,3))
arr = np.arange(10, 19, 1)
print(f"list: {list}")
print(f"np.array: {array}")
print(f"np.zeros((7,)): {zeros}")
print(f"np.ones((2,3)): {ones}")
print(f"np.arange(10, 19, 1): {arr}")
```

**2. Базові операції**

Створіть Numpy-масив зі значеннями [12, 24, 36, 48, 60].

Виконайте наступні операції над **оригінальним** масивом і запишіть у нову змінну результат (кожен пункт оцінюється в 0.5 бала):
- Помножте всі елементи на 2
- Додайте до всіх елементів 10
- Піднесіть всі елементи до квадрату
- Обчисліть квадратний корінь з усіх елементів

Виведіть результат виконання кожної операції.

```python
import numpy as np

array = np.array([12, 24, 36, 48, 60])

array_multiplied_by_2 = np.multiply(array, 2) # or array_multiplied_by_2 = array * 2
array_plus_10 = np.add(array, 10) # or array_plus_10 = array + 10
array_squared = np.square(array) # or array_squared = array ** 2
array_sqrt = np.sqrt(array)

print(f"array multiplied by 2: {array_multiplied_by_2}")
print(f"array plus 10: {array_plus_10}")
print(f"array squared: {array_squared}")
print(f"array square root: {array_sqrt}")
```

**3. Статистичні функції**

Створіть Numpy-масив `data` зі значеннями [18, 27, 33, 41, 52, 68, 79].

Обчисліть використовуючи функціонал numpy (кожен пункт оцінюється в 0.5 бала):
- Середнє значення
- Мінімум
- Максимум
- Суму всіх елементів

```python
import numpy as np

data =np.array([18, 27, 33, 41, 52, 68, 79])

mean = np.mean(data)
min_value = np.min(data)
max_value = np.max(data)
sum = np.sum(data)

print(f"Mean: {mean}")
print(f"Min: {min_value}")
print(f"Max: {max_value}")
print(f"Sum: {sum}")
```

**4. Індексація**

Створіть Numpy-масив `arr` зі значеннями [15, 22, 35, 44, 58, 67].

Виведіть (кожен пункт оцінюється в 0.5 бала):
- Перший елемент
- Останній елемент
- Елементи з індексами 1 до 3
- Елементи більші за 40

```python
import numpy as np

arr = np.array([15, 22, 35, 44, 58, 67])

print(f"Первый элемент: {arr[0]}")
print(f"Последний элемент: {arr[-1]}")
print(f"Интервал: {arr[1:4]}")

mask = arr > 40
print(f"Элементы, большие 40: {arr[mask]}")
```

**5. Двовимірні масиви**

Задано матрицю 3x3.

Виконайте наступні дії з функціоналом Numpy (кожен пункт оцінюється в 0.5 бала):
- Виведіть форму (shape) матриці
- Виведіть перший рядок
- Виведіть другу колонку
- Обчисліть суму всіх елементів
- Обчисліть середнє всіх елементів

```python
import numpy as np

matrix = np.array([[2, 4, 6],
                   [8, 10, 12],
                   [14, 16, 18]])

print(f"Форма матриці: {matrix.shape}")
print(f"Перший рядок: {matrix[0]}")
print(f"Другий стовпець: {matrix[:, 1]}")

print(f"Сума елементів матриці: {np.sum(matrix)}")
print(f"Середнє значення елементів матриці: {np.mean(matrix)}")
```
